```python
import pandas as pd
pd.set_option('display.float_format', '{:.1e}'.format)
```

```python
# read the stored data

data = pd.read_csv("data-2.csv")
```

```python
def top_5(data, key):
    data_abs = data.abs()
    return data_abs.sort_values(by=[key],ascending=False)[['muTilde', 'mqt3', 'mut3', 'mdt3', 'cH', 'cHD', 'cHBox', 'cHG', 'cuH_33', 'cHu_33', 'cHq1_33', 'cHq3_33', 'cuG_33', 'cuu_3333', 'cqu8_3333','cqd8_3333', 'cud8_3333']].head()
```

 Locate mass-combinations that give largest values for various coefficients

```python
coeff_list = ['cH', 'cHD', 'cHBox', 'cHG', 'cuH_33', 'cHu_33', 'cHq1_33', 'cHq3_33', 'cuG_33', 'cuu_3333', 'cqu8_3333','cqd8_3333', 'cud8_3333']
columns = ['muTilde', 'mqt3', 'mdt3', 'mut3'] + coeff_list

row_nums = []
for coeff in coeff_list:
    row_nums += list(top_5(data,coeff).index)

rows = list(set(row_nums))
rows
```

```python
data.loc[rows,columns].sort_values(by='cH', ascending=True)
```
