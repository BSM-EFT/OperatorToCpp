```python
import pandas as pd
pd.set_option('display.float_format', '{:.1e}'.format)
```

Read the stored data

```python
data = pd.read_csv("data-2.csv")
data['cHWB'].min()
```

 Locate mass-combinations that give largest values for various coefficients

```python
coeff_list = ['cHD','cHG','cHu_33', 'cHq1_33', 'cHq3_33', 'cuG_33']
columns = ['muTilde', 'mqt3', 'mdt3', 'mut3'] + coeff_list
```

```python
def top_5(data, key):
    data_abs = data.abs()
    return data_abs.sort_values(by=[key],ascending=False)[columns].head()
```

```python
row_nums = []
for coeff in coeff_list:
    row_nums += list(top_5(data,coeff).index)

rows = list(set(row_nums))
rows
```

```python
data.loc[rows,columns].sort_values(by='cHq1_33', ascending=False)
```
